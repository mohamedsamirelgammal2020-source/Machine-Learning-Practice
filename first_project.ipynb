```python
import pandas as pd
from sklearn.ensemble import RandomForestRegressor

cars_data = {
    'Car_Model': ['VW ID.4', 'Panamera Turbo', 'Model S', 'Taycan'],
    'HorsePower': [201, 550, 670, 402],
    'Cabin_Space_cm': [104, 99, 100, 98],
    'Price_AED': [170000, 700000, 350000, 400000]
}

df = pd.DataFrame(cars_data)
print("البيانات التاريخية للسيارات:")
print(df)
print("-" * 40)

X = df[['HorsePower', 'Cabin_Space_cm']]
y = df['Price_AED']

ai_model = RandomForestRegressor(n_estimators=100, random_state=42)
ai_model.fit(X, y)

new_car_specs = pd.DataFrame([[450, 102]], columns=['HorsePower', 'Cabin_Space_cm'])

predicted_price = ai_model.predict(new_car_specs)

print(f"🔥 الذكاء الاصطناعي يتوقع أن سعر السيارة الجديدة سيكون: {predicted_price[0]:,.2f} درهم")
predicted_price = ai_model.predict(new_car_specs)

print(f"🔥 الذكاء الاصطناعي يتوقع أن سعر السيارة الجديدة سيكون: {predicted_price[0]:,.2f} درهم")
```

```python
import matplotlib.pyplot as plt

models = ['VW ID.4', 'Model S', 'Taycan', 'Panamera Turbo', 'New AI Car']

prices = [170000, 350000, 400000, 700000, predicted_price[0]]

plt.figure(figsize=(10, 5))
plt.bar(models, prices, color=['skyblue', 'lightgreen', 'orange', 'red', 'purple'])
plt.title('Car Prices Comparison (AED)', fontsize=14)
plt.ylabel('Price (AED)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```

```python
my_budget = 1000000

print("هل نقدر نشتري السيارة اللي تنبأنا بسعرها؟ 🤔")
print("-" * 40)

if predicted_price[0] <= my_budget:
    print("يا سلام! الميزانية تكفي، مبروك الموتر الجديد! 🏎️💨")
else:
    shortage = predicted_price[0] - my_budget
    print(f"أوف! للأسف الميزانية ما تكفي.. ناقصنا {shortage:,.2f} درهم.")
    print (f"لا تشتريها طلعت اغلى من الميزانية")
```
